#### 6.- Realiza una función que realice una digestión con las siguientes enzimas de restricción en nuestra secuencia. La función tiene que devolver la siguiente información: (2 pto)
1. Enzimas que tienen sitios de corte en la secuencia y posición (o posiciones de corte)
2. Enzimas que no tienen sitios de corte (si las hay)
3. Resultado de la digestión: los fragmentos resultantes de nuestra secuencia
4. Indicar número y tamaño de cada fragmento

```python
# Diccionario con las enzimas y su secuencia de reconocimiento
enzR = {
    'HindIII' : 'A*AGCTT',
    'BamHI'   : 'G*GATCC',
    'AluI'    : 'AG*CT',
    'Sau3AI'  : '*GATC',
    'EcoRI'   : 'G*AATTC'
}
```

```python

def digestion(secuencia, enzR):
    # Nos aseguramos de trabajar en mayúsculas
    secuencia = secuencia.upper()
    
    enzimas_con_corte = {}   # {enzima: [posiciones_de_corte]}
    enzimas_sin_corte = []   # [enzima, ...]
    todas_pos_corte = []     # todas las posiciones de corte de todas las enzimas

    # 1) Buscar sitios de corte de cada enzima
    for enzima, sitio in enzR.items():
        # Posición del asterisco dentro del sitio (dónde corta)
        idx_corte = sitio.index('*')
        # Patrón real de ADN sin el asterisco
        patron = sitio.replace('*', '')
        
        posiciones = []
        start = 0
        
        # Buscamos todas las apariciones del patrón
        while True:
            pos = secuencia.find(patron, start)
            if pos == -1:
                break
            
            # Posición de corte real en la secuencia
            pos_corte = pos + idx_corte      # índice 0-based
            posiciones.append(pos_corte)
            todas_pos_corte.append(pos_corte)
            
            start = pos + 1  # seguimos buscando a partir del siguiente nucleótido
        
        if posiciones:
            enzimas_con_corte[enzima] = posiciones
        else:
            enzimas_sin_corte.append(enzima)

    # 2) Resultado de la digestión: fragmentos
    # Ordenamos y quitamos duplicados de las posiciones de corte
    todas_pos_corte = sorted(set(todas_pos_corte))
    
    fragmentos = []
    prev = 0
    for pc in todas_pos_corte:
        fragmentos.append(secuencia[prev:pc])
        prev = pc
    # Último fragmento desde el último corte hasta el final
    fragmentos.append(secuencia[prev:])
    
    # 3) Número y tamaño de cada fragmento
    tamanos_fragmentos = [len(frag) for frag in fragmentos]
    
    # Devolvemos toda la información
    return enzimas_con_corte, enzimas_sin_corte, fragmentos, tamanos_fragmentos

# ===== Ejemplo de uso (suponiendo que ya tienes definida 'secuencia') =====
enz_cortan, enz_no_cortan, frags, tamaños = digestion(secuencia, enzR)

print("1) Enzimas que cortan y posiciones de corte:")
for e, pos in enz_cortan.items():
    if len(pos) <= 10:
        print(f"{e}: {pos}")
    else:
        print(f"{e}: {pos[:10]} ...")

print("\n2) Enzimas que no cortan:")
print("   ", enz_no_cortan)

print("\n3) Fragmentos resultantes de la digestión:")
# Para mostrar los 10 primeros fragmentos
maximo = 10
for i, frag in enumerate(frags[:maximo], start=1):
    print(f"   Fragmento {i}: {frag[:30]}{'...' if len(frag)>30 else ''}")
    # Si hay más de 10, añadimos "..."
if len(frags) > maximo:
        print("  ...")

print("\n4) Número y tamaño de cada fragmento:")
num_frag = len(frags)
print(f"   Nº de fragmentos: {len(frags)}")

if num_frag <= 10:
    print("Tamaños de los fragmentos:", tamaños)
else:
    print("   Tamaños de los fragmentos:", tamaños[:10], "...")
```

#### 7.- Realiza una función que mediante la introducción de una secuencia de ADN devuelva lo siguiente: (3pto)
1. Todos los ORFs posibles que puedan surgir de la secuencia.
2. Selecciona los 6 ORFs más largos e indica si corresponden a la strand positiva o negativa.
3. Devuelve como mejor predicción, el más largo de todos.
4. Calcula el porcentaje de aminoácidos aromáticos de esa proteína elegida y su longitud relativa a la secuencia original de ADN.

```python
# Tabla de codones ADN -> aminoácidos (una letra)
tabla_codon = {
    'TTT':'F','TTC':'F','TTA':'L','TTG':'L',
    'CTT':'L','CTC':'L','CTA':'L','CTG':'L',
    'ATT':'I','ATC':'I','ATA':'I','ATG':'M',
    'GTT':'V','GTC':'V','GTA':'V','GTG':'V',
    'TCT':'S','TCC':'S','TCA':'S','TCG':'S',
    'CCT':'P','CCC':'P','CCA':'P','CCG':'P',
    'ACT':'T','ACC':'T','ACA':'T','ACG':'T',
    'GCT':'A','GCC':'A','GCA':'A','GCG':'A',
    'TAT':'Y','TAC':'Y','TAA':'*','TAG':'*',
    'CAT':'H','CAC':'H','CAA':'Q','CAG':'Q',
    'AAT':'N','AAC':'N','AAA':'K','AAG':'K',
    'GAT':'D','GAC':'D','GAA':'E','GAG':'E',
    'TGT':'C','TGC':'C','TGA':'*','TGG':'W',
    'CGT':'R','CGC':'R','CGA':'R','CGG':'R',
    'AGT':'S','AGC':'S','AGA':'R','AGG':'R',
    'GGT':'G','GGC':'G','GGA':'G','GGG':'G'
}

# Función para traducir una secuencia de ADN a proteína
def traducir_dna_a_proteina(seq_dna):
    seq_dna = seq_dna.upper()
    proteina = ""
    
    # Recorro de 3 en 3 (codones)
    for i in range(0, len(seq_dna) - 2, 3):
        codon = seq_dna[i:i+3]
        if codon in tabla_codon:
            aa = tabla_codon[codon]
        else:
            aa = 'X'   # por si aparece algo raro
        proteina += aa
    
    return proteina
```

```python
# Complementos en ADN
complementos_dna = {
    'A': 'T',
    'T': 'A',
    'C': 'G',
    'G': 'C'
}

def complemento_inverso(seq):
    seq = seq.upper()
    comp = ""
    
    # Primero complemento
    for base in seq:
        if base in complementos_dna:
            comp += complementos_dna[base]
        else:
            comp += 'N'
    
    # Luego le doy la vuelta
    comp_rev = comp[::-1]
    return comp_rev
```

```python
def encontrar_orfs_en_strand(seq, strand):
    seq = seq.upper()
    start_codon = 'ATG'
    stop_codones = ['TAA', 'TAG', 'TGA']
    
    lista_orfs = []
    
    # Recorro los 3 frames: 0, 1 y 2
    for frame in range(3):
        i = frame
        while i < len(seq) - 2:
            codon = seq[i:i+3]
            
            if codon == start_codon:
                inicio = i
                j = i
                # Busco el STOP
                while j < len(seq) - 2:
                    codon2 = seq[j:j+3]
                    if codon2 in stop_codones:
                        fin = j + 3    # incluyo el STOP
                        orf_dna = seq[inicio:fin]
                        
                        # Guardo info en un diccionario
                        info_orf = {
                            'strand': strand,
                            'frame': frame,
                            'inicio': inicio,
                            'fin': fin,
                            'dna': orf_dna
                        }
                        lista_orfs.append(info_orf)
                        break
                    j += 3
                i = j
            else:
                i += 3
    
    return lista_orfs
```
